## 4. Inspect the attached dataset

When running this notebook on Kaggle, first attach **Amir Mahdi Zeynali - Machine Learning Portfolio** using **Add Input**. The code below looks for the ZIP file in common Kaggle input locations and reports its contents without extracting the image collection. If the dataset is not attached as a ZIP, it prints the available input entries so you can identify its layout.


```python
from pathlib import Path
from collections import Counter
import zipfile

SEARCH_ROOTS = [Path("/kaggle/input"), Path.cwd()]
archive_candidates = []

for root in SEARCH_ROOTS:
    if not root.exists():
        continue
    # Check the root and its immediate child folders to avoid an expensive
    # recursive walk through tens of thousands of image files.
    for candidate in [root, *[p for p in root.iterdir() if p.is_dir()]]:
        if candidate.is_dir():
            archive_candidates.extend(
                p for p in candidate.iterdir()
                if p.is_file() and p.suffix.lower() == ".zip"
            )

# Prefer the archive name used in the original portfolio.
archive_candidates = list(dict.fromkeys(archive_candidates))
archive_path = next(
    (p for p in archive_candidates if p.name.lower() in {"رزومه.zip", "amir_mahdi_zeynali_ml_portfolio.zip"}),
    archive_candidates[0] if archive_candidates else None,
)

if archive_path is None:
    print("No ZIP archive was found in the common input locations.")
    for root in SEARCH_ROOTS:
        if root.exists():
            print(f"\nContents of {root}:")
            for item in list(root.iterdir())[:30]:
                print(" -", item.name, "/" if item.is_dir() else "")
else:
    print("Archive:", archive_path)
    print(f"Size: {archive_path.stat().st_size / (1024**2):.1f} MiB")
    with zipfile.ZipFile(archive_path) as zf:
        files = [name for name in zf.namelist() if not name.endswith("/")]
        extensions = Counter(Path(name).suffix.lower() or "[no extension]" for name in files)
        print("Files in archive:", len(files))
        print("Jupyter notebooks:", extensions[".ipynb"])
        print("CSV files:", extensions[".csv"])
        print("JPG images:", extensions[".jpg"] + extensions[".jpeg"])

        class_prefixes = [
            "رزومه/cnn/Training/female/",
            "رزومه/cnn/Training/male/",
            "رزومه/cnn/Validation/female/",
            "رزومه/cnn/Validation/male/",
        ]
        for prefix in class_prefixes:
            count = sum(name.startswith(prefix) for name in files)
            print(f"{prefix}: {count:,} files")
```
